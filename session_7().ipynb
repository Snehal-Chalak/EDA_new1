### Z score

## z= x- u / sigma

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

path="visadataset.csv"
visa_df=pd.read_csv(path)
visa_df


cat=visa_df.select_dtypes(include='object').columns
num=visa_df.select_dtypes(exclude='object').columns
```

```python
 # step 1: take the pwage data
# 2: calulate mean of the data
# calcualte std of the data
# 4 Nr=data-mean step1 - step2
# 5 z_Wage=Nr/std=Step4/step3
# 6 create a new column visa_df['z_wage']=z_wage
```

```python
data=visa_df['prevailing_wage']

mean=data.mean()
std=data.std()
Nr=data-mean
z_score=Nr/std
#visa_df['prevailing_wage_z']=(wage_data-wage_mean)/wage_std

visa_df['z_score']=z_score
#visa_df[['prevailing_wage','prevailing_wage_z']]
```

### Min_Max Scaler

Xscaler=X - Xmin / Xmax - Xmin

1 reade data
2 calculate min
3 calculate max
Nr=step1- step2
Dr=step3- step2
Nr/Dr
visa_df['wage_min_max']=Nr/Dr

```python
wage_data=visa_df['prevailing_wage']
wage_Min=wage_data.min()
wage_Max=wage_data.max()
Nr=wage_data-wage_Min
Dr=wage_Max - wage_Min
x_scaled = Nr / Dr
print(x_scaled)
#visa_df=[]
```
