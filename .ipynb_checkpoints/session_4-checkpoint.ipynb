```python
# step 1:- calulet q1
# step 2:- calulet q2
# step 3:- calulet q3
# step 4:- iqr=q3-q1
# step 5:- lb= q1-1.5*iqr
# step 6:- ub=q3+1.5*iqr
# step 7:- con1
# step 8:- con2
# step 9:- con
# step 10:- get the count
```

```python
q1=np.percentile(wage_data,25)

q2=np.percentile(wage_data,50)

q3=np.percentile(wage_data,75)

IQR= q3-q1

LB= q1-1.5*IQR
UB= q3+1.5*IQR

con1=wage_data<LB
con2=wage_data>UB
con=con1 | con2

outliers_df=visa_df[con]
outliers_df['prevailing_wage']
outliers_df['prevailing_wage'].values
outliers_list=outliers_df['prevailing_wage'].values.tolist()
```

```python
outliers_list=outliers_df['prevailing_wage'].values.tolist()
```

```python
# I want to call the each observetion of wegw data
# if that observation will avaible in this 427 values
# that observations will replace by media
# otherwise thet observatoin will be avalable as it is

wage_median = visa_df['prevailing_wage'].median()
wage_data = visa_df['prevailing_wage']
new_wage_data=[]
for i in wage_data:
    if i in outliers_list:
        new_wage_data.append(wage_median)
    else:
          new_wage_data.append(i)
```
